Αρχικά κατεβάζουμε από το διαδίκτυο τα αναγκαία αρχεία CSV.

(Πηγή δεδομένων https://www.kaggle.com/datasets/rinichristy/covid19-coronavirus-pandemic)

```python
# Πρώτο αρχείο
remote_url = 'https://raw.githubusercontent.com/kokkytos/programming/main/docs/COVID-19%20Coronavirus.csv'
# ορισμός του ονόματος αρχείου για τοπική αποθήκευση
local_file1 = 'COVID-19_Coronavirus.csv'
# Λήψη και αποθήκευση τοπικά
request.urlretrieve(remote_url, local_file1)

# Δεύτερο αρχείο
remote_url = 'https://raw.githubusercontent.com/kokkytos/programming/main/docs/COVID-19%20Coronavirus_V2.csv'
# ορισμός του ονόματος αρχείου για τοπική αποθήκευση
local_file2 = 'COVID-19_Coronavirus_V2.csv'
# Λήψη και αποθήκευση τοπικά
request.urlretrieve(remote_url, local_file2)
```

Ακολουθεί η ανάγνωση του πρώτου αρχείου με την χρήση του `csv.reader`. 
Ο παρακατω κώδικας διαβάζει το αρχείο CSV γραμμή προς γραμμή και τυπώνει κάθε γραμμή ως λίστα από συμβολοσειρές (strings).

Ο `reader` δημιουργεί έναν αναγνώστη CSV (`csv.reader`) πάνω στο ανοιχτό αρχείο. Διαχειρίζεται σωστά τα κόμματα (ή άλλους διαχωριστές) και τα εισαγωγικά μέσα στα πεδία.

```python
with open(local_file1, 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

Το επόμενη κελί κώδικα ξεχωρίζει την πρώτη γραμμή (τις επικεφαλίδες/ονόματα στηλών) από τις υπόλοιπες γραμμές δεδομένων και την μορφοποιεί διαφορετικά κατά την εκτύπωση.

```python
with open(local_file1, 'r') as file:
    reader = csv.reader(file)
    line_count = 0
    for row in reader:
        if line_count == 0:
            print(f'Ονόματα στηλών {", ".join(row)}')
            print("\n\n")
            line_count += 1
        else:
            print(row)
```

To κόμμα (,) χρησιμοποιείται ως προκαθορισμένη επιλογή σαν διαχωριστικό στηλών.
Μπορούμε ρητά να δηλώσουμε ποιο θα είναι το διαχωριστικό με την επιλογή _delimiter_.

```python
with open(local_file2, 'r') as file:
    reader = csv.reader(file, delimiter = ';',quoting=csv.QUOTE_ALL)
    for row in reader:
        print(row)
```

Ένας άλλος τρόπος ανάγνωσης είναι μέσω της `csv.DictReader()`.

Η `csv.DictReader()` διαβάζει το CSV και μετατρέπει κάθε γραμμή του σε λεξικό (dictionary).

Συγκεκριμένα:
- Η πρώτη γραμμή του CSV θεωρείται ότι περιέχει τις επικεφαλίδες (headers).
- Οι επικεφαλίδες γίνονται τα keys του λεξικού.
- Οι τιμές κάθε γραμμής γίνονται τα αντίστοιχα values.

```python
with open(local_file1, 'r') as file:
    csv_file = csv.DictReader(file)
    for row in csv_file:
        print(dict(row))
```

## Εγγραφή σε csv

Ο επόμενος κώδικας δημιουργεί ένα αρχείο CSV με όνομα employee_file.csv και γράφει μέσα σε αυτό τα στοιχεία δύο εργαζομένων.



```python
with open('employee_file.csv', mode='w') as employee_file:
    employee_writer = csv.writer(employee_file, delimiter=';', quotechar='"', quoting=csv.QUOTE_ALL)

    employee_writer.writerow(['John Smith', 'Accounting', 'June'])
    employee_writer.writerow(['Erica Meyers', 'IT', 'March'])
```

Ας τον δούμε πιο αναλυτικά:

Η `csv.writer()` δημιουργεί ένα αντικείμενο που μας επιτρέπει να γράφουμε δεδομένα στο CSV αρχείο.

Ας δούμε τις παραμέτρους:

- α) employee_file

Το αρχείο στο οποίο θα γράψουμε τα δεδομένα.

- β) delimiter=';'

Καθορίζει τον χαρακτήρα που διαχωρίζει τις στήλες.

Εδώ χρησιμοποιούμε το ελληνικό ερωτηματικό (;) αντί για το κόμμα (,).

- γ) quotechar='"'

Καθορίζει τον χαρακτήρα που χρησιμοποιείται για να περικλείει τις τιμές κάθε πεδίου.

Εδώ χρησιμοποιούμε τα διπλά εισαγωγικά (").

- δ) quoting=csv.QUOTE_ALL

Καθορίζει ότι όλες οι τιμές θα περικλείονται υποχρεωτικά σε διπλά εισαγωγικά.
Ακόμα και αν μια τιμή δεν χρειάζεται εισαγωγικά, θα τα χρησιμοποιήσει.

Η `writerow()` γράφει μία γραμμή στο CSV αρχείο.

Εγγραφή σε CSV από dictionary (λεξικό). Σε αυτή την περίπτωση πρέπει να ορίσουμε προκαταβολικά τα ονόματα των στηλών (fieldnames παράμετρος στην μέθοδο DictWriter) για να γίνει η ταυτοποίηση των κλειδιών από το λεξικό κατά την εγγραφή.

```python
with open('employee_file2.csv', mode='w') as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames= ['emp_name', 'dept', 'birth_month'],delimiter=';')

    writer.writeheader()
    writer.writerow({'emp_name': 'John Smith', 'dept': 'Accounting', 'birth_month': 'November'})
    writer.writerow({'emp_name': 'Erica Meyers', 'dept': 'IT', 'birth_month': 'March'})
```

Διαφορά της `writerow()` από την `writerows()`

- `writerow()`: Γράφει μία γραμμή κάθε φορά.
- `writerows()`: Γράφει πολλές γραμμές μαζί. Τα δεδομένα για κάθε γραμμή είναι ένα λεξικό. Οπότε για όλες τις γραμμές το `writerows()` δέχεται μια λίστα από λεξικά.
Για παράδειγμα, μπορούμε να γράψουμε δεδομένα με τον παρακάτω τρόπο:


```python
with open('employee_file2.csv', mode='w') as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames= ['emp_name', 'dept', 'birth_month'],delimiter=';')
    writer.writeheader()
    writer.writerows([
        {'emp_name': 'John Smith',
         'dept': 'Accounting',
         'birth_month': 'June'},

        {'emp_name': 'Erica Meyers',
         'dept': 'IT',
         'birth_month': 'March'}
    ])
```

Μετατροπή της στήλης last_review σε datetime64 object

```python
airbnb['last_review'] = pd.to_datetime(airbnb['last_review'])
airbnb.dtypes
```

Μπορούμε να εξάγουμε το έτος από ένα datetime object σε μια νέα στήλη. Μετατροπή της στήλης year σε ακέραιο
